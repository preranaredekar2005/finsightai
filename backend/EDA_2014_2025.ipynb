```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display all columns
pd.set_option("display.max_columns", None)

# Better looking plots
plt.style.use("ggplot")

print("Libraries imported successfully!")
```

```python
# ================================
# Load CSV Files
# ================================

stock = pd.read_csv("models/saved/stock_prices_2014_2025.csv")
technical = pd.read_csv("models/saved/technical_indicators_2014_2025.csv")
news = pd.read_csv("models/saved/news_articles_2014_2025.csv")
merged = pd.read_csv("models/saved/merged_market_dataset_2014_2025.csv")

print("="*50)
print("Datasets Loaded Successfully")
print("="*50)

print(f"Stock Prices           : {stock.shape}")
print(f"Technical Indicators   : {technical.shape}")
print(f"News Articles          : {news.shape}")
print(f"Merged Dataset         : {merged.shape}")
```

```python
# ==========================================
# Dataset Overview
# ==========================================

print("="*60)
print("STOCK PRICES DATASET")
print("="*60)

display(stock.head())

print("\nShape :", stock.shape)
print("\nColumns :")
print(stock.columns.tolist())


print("\n")
print("="*60)
print("TECHNICAL INDICATORS DATASET")
print("="*60)

display(technical.head())

print("\nShape :", technical.shape)
print("\nColumns :")
print(technical.columns.tolist())


print("\n")
print("="*60)
print("NEWS ARTICLES DATASET")
print("="*60)

display(news.head())

print("\nShape :", news.shape)
print("\nColumns :")
print(news.columns.tolist())


print("\n")
print("="*60)
print("MERGED DATASET")
print("="*60)

display(merged.head())

print("\nShape :", merged.shape)
print("\nColumns :")
print(merged.columns.tolist())
```

```python
# ==========================================
# Missing Values Analysis
# ==========================================

print("="*60)
print("Missing Values")
print("="*60)

missing = merged.isnull().sum()

missing = missing[missing > 0]

if len(missing) == 0:
    print("No Missing Values Found ✅")
else:
    display(missing.sort_values(ascending=False))
```

```python
# ==========================================
# Duplicate Records
# ==========================================

duplicates = merged.duplicated().sum()

print("="*60)
print("Duplicate Records")
print("="*60)

print(f"Total Duplicate Rows : {duplicates}")
```

```python
# ==========================================
# Dataset Information
# ==========================================

merged.info()
```

```python
# ==========================================
# Closing Price Distribution
# ==========================================

plt.figure(figsize=(10,5))

plt.hist(stock["close"], bins=50)

plt.title("Distribution of Closing Prices")
plt.xlabel("Closing Price")
plt.ylabel("Frequency")

plt.show()
```

```python
# ==========================================
# Stock Closing Price Trend
# ==========================================

stock["date"] = pd.to_datetime(stock["date"])

daily_close = stock.groupby("date")["close"].mean()

plt.figure(figsize=(15,6))

plt.plot(daily_close)

plt.title("Average Closing Price (2014-2025)")
plt.xlabel("Year")
plt.ylabel("Average Closing Price")

plt.show()
```

```python
# ==========================================
# Trading Volume Distribution
# ==========================================

plt.figure(figsize=(10,5))

plt.hist(stock["volume"], bins=50)

plt.title("Trading Volume Distribution")
plt.xlabel("Volume")
plt.ylabel("Frequency")

plt.show()
```

```python
# ==========================================
# Top Stocks
# ==========================================

top = stock.groupby("ticker")["close"].mean().sort_values(ascending=False)

plt.figure(figsize=(12,6))

top.head(10).plot(kind="bar")

plt.title("Top 10 Stocks by Average Closing Price")
plt.xlabel("Ticker")
plt.ylabel("Average Close")

plt.show()
```

```python
# ==========================================
# Correlation Heatmap
# ==========================================

numeric = merged.select_dtypes(include=np.number)

corr = numeric.corr()

plt.figure(figsize=(14,10))

sns.heatmap(
    corr,
    cmap="coolwarm",
    annot=False
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
# ==========================================
# RSI Distribution
# ==========================================

plt.figure(figsize=(10,5))

plt.hist(technical["rsi"], bins=40)

plt.title("RSI Distribution")
plt.xlabel("RSI")
plt.ylabel("Frequency")

plt.show()
```

Most RSI values lie between 40 and 70, indicating that the market generally remained in a neutral trading zone, with relatively fewer instances of oversold (<30) or overbought (>70) conditions.

```python
# ==========================================
# MACD Distribution
# ==========================================

plt.figure(figsize=(10,5))

plt.hist(technical["macd"], bins=40)

plt.title("MACD Distribution")
plt.xlabel("MACD")
plt.ylabel("Frequency")

plt.show()
```

The MACD values are centered around zero, indicating a balanced mix of bullish and bearish momentum throughout the dataset.

```python
# ==========================================
# Bollinger Band Percentage
# ==========================================

plt.figure(figsize=(10,5))

plt.hist(technical["bb_pct"], bins=40)

plt.title("Bollinger Band Percentage")
plt.xlabel("BB %")
plt.ylabel("Frequency")

plt.show()
```

Most observations lie within the Bollinger Bands, suggesting normal market volatility with occasional breakout periods.

```python
# ==========================================
# Sentiment Distribution
# ==========================================

plt.figure(figsize=(7,5))

news["sentiment_label"].value_counts().plot(kind="bar")

plt.title("News Sentiment Distribution")
plt.xlabel("Sentiment")
plt.ylabel("Count")

plt.show()
```

The news dataset contains positive, neutral, and negative sentiment labels, providing additional market context that complements technical indicators.

```python
# ==========================================
# News Sources
# ==========================================

plt.figure(figsize=(10,5))

news["source"].value_counts().head(10).plot(kind="bar")

plt.title("Top News Sources")
plt.xlabel("Source")
plt.ylabel("Articles")

plt.show()
```

```python
# ==========================================
# Records per Stock
# ==========================================

plt.figure(figsize=(12,6))

stock["ticker"].value_counts().sort_values().plot(kind="barh")

plt.title("Records Available for Each Ticker")
plt.xlabel("Number of Records")
plt.ylabel("Ticker")

plt.show()
```

```python
# ==========================================
# Average Closing Price by Year
# ==========================================

stock["Year"] = pd.to_datetime(stock["date"]).dt.year

yearly = stock.groupby("Year")["close"].mean()

plt.figure(figsize=(10,5))

plt.plot(yearly.index, yearly.values, marker="o")

plt.title("Average Closing Price by Year")
plt.xlabel("Year")
plt.ylabel("Average Close")

plt.grid(True)

plt.show()
```

The historical data from 2014–2025 captures long-term market movements and provides a robust foundation for training predictive models.
